1) What is the difference between descriptive statistics and inferential
statistics? Explain with examples.
- Descriptive Statistics : Descriptive statistics are used to summarize, organize, and describe the data you already have.They do not make predictions or conclusions beyond the given data.
- Inferential statistics : Inferential statistics use a sample of data to make conclusions or predictions about a larger population. They go beyond the available data.
- |Basis|Descriptive Statistics|Inferential Statistics|
|-------------------------------|------------------|-------------------------|
|Purpose|To present data in a meaningful way|To make predictions or generalizations|
|Type of data used|Uses the complete dataset|Uses a sample from the population|
|Scope|Limited to the observed data only|Extends beyond the observed data|
|Decision making|Does not support decision making|Helps in decision making|
|Techniques used|Mean, Median, Mode, Range, Standard Deviation, Graphs|Hypothesis testing, Confidence interval, Regression, ANOVA|
|Prediction|No prediction involved|Used for prediction and estimation|
|Examples|Average marks of a class |Predicting average marks of all students from a sample|

2) What is sampling in statistics? Explain the differences between random
and stratified sampling.
- Sampling is the process of selecting a subset (sample) from a larger group (population) to draw conclusions about the whole population.
- Studying a sample is easier, cheaper, and faster than studying the entire population.
- Example : Instead of surveying all voters in a city, you survey 1,000 voters and use the results to understand the opinions of the entire city.
- |Basis|Random Sampling|Stratified Sampling|
|------------------------|-----------------------|--------------------------|
|Meaning|Selection is completely by chance|Population is divided into strata before sampling|
|Population division|No division of population|Population divided into homogeneous groups|
|Chance of selection|Equal chance for all individuals|Equal chance within each stratum|
|Representation|Some groups may be underrepresented|All groups are represented|
|Accuracy|Less accurate for diverse populations|More accurate and reliable|
|Complexity|Simple and easy|More complex|
|Example|Lottery method|Sampling boys and girls separately|

4) Explain skewness and kurtosis. What does a positive skew imply about the
data?
- Skewness : measures the degree of asymmetry of a frequency distribution about its mean.
- If the distribution is symmetrical → skewness = 0
- If it is stretched more on one side → skewness ≠ 0
- Types of skewness:
- Positive skewness (right-skewed): Long tail on the right
- Negative skewness (left-skewed): Long tail on the left
- Kurtosis : measures the peakedness or flatness of a distribution compared to a normal distribution.
- It indicates how concentrated the values are around the mean and how heavy the tails are.
- Types of kurtosis:
- a) Leptokurtic - Sharp peak, heavy tails
- b) Mesokurtic - Normal peak (normal distribution)
- c) Platykurtic - Flat peak, light tails
- A positive skew implies that:
- 1) The distribution has a long right tail
- 2) Most observations are concentrated at lower values
- 3) A few large values (outliers) pull the mean to the right
- 4) Relationship:Mean>Median>Mode
- Example:Income distribution where most people earn low to moderate income, but a few earn very high income.

5) Implement a Python program to compute the mean, median, and mode of
a given list of numbers.
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]



```python
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Mean
mean = sum(numbers) / len(numbers)

# Median
numbers_sorted = sorted(numbers)
n = len(numbers_sorted)
if n % 2 == 0:
    median = (numbers_sorted[n//2 - 1] + numbers_sorted[n//2]) / 2
else:
    median = numbers_sorted[n//2]

# Mode
frequency = {}
for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1

max_freq = max(frequency.values())
mode = [key for key, value in frequency.items() if value == max_freq]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

6)  Compute the covariance and correlation coefficient between the following
two datasets provided as lists in Python:
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

```python
import math

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

n = len(list_x)

# Means
mean_x = sum(list_x) / n
mean_y = sum(list_y) / n

# Covariance (sample)
covariance = sum((list_x[i] - mean_x) * (list_y[i] - mean_y) for i in range(n)) / (n - 1)

# Standard deviations (sample)
std_x = math.sqrt(sum((x - mean_x) ** 2 for x in list_x) / (n - 1))
std_y = math.sqrt(sum((y - mean_y) ** 2 for y in list_y) / (n - 1))

# Correlation coefficient
correlation = covariance / (std_x * std_y)

print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

7) Write a Python script to draw a boxplot for the following numeric list and
identify its outliers. Explain the result:
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

```python
import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Draw boxplot
plt.boxplot(data)
plt.title("Boxplot of Given Data")
plt.ylabel("Values")
plt.show()

# Outlier detection using IQR method
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Outliers:", outliers)
```

You are working as a data analyst in an e-commerce company. The
marketing team wants to know if there is a relationship between advertising spend and
daily sales.
● Explain how you would use covariance and correlation to explore this
relationship.
● Write Python code to compute the correlation between the two lists:
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

- Covariance: is used to check the direction of the relationship between advertising spend and daily sales.
- If covariance is positive, higher advertising spend is associated with higher sales.
- If covariance is negative, higher advertising spend is associated with lower sales.
- Correlation : is used to measure the strength and direction of the relationship.
- Correlation values range from -1 to +1
- A value close to +1 indicates a strong positive relationship.
- In this scenario, we compute correlation to clearly understand how strongly advertising spend affects daily sales.

```python
import math

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

n = len(advertising_spend)

mean_x = sum(advertising_spend) / n
mean_y = sum(daily_sales) / n

covariance = sum(
    (advertising_spend[i] - mean_x) * (daily_sales[i] - mean_y)
    for i in range(n)
) / (n - 1)

std_x = math.sqrt(sum((x - mean_x) ** 2 for x in advertising_spend) / (n - 1))
std_y = math.sqrt(sum((y - mean_y) ** 2 for y in daily_sales) / (n - 1))

correlation = covariance / (std_x * std_y)

print("Correlation Coefficient:", correlation)
```

9)  Your team has collected customer satisfaction survey data on a scale of
1-10 and wants to understand its distribution before launching a new product.
● Explain which summary statistics and visualizations (e.g. mean, standard
deviation, histogram) you’d use.
● Write Python code to create a histogram using Matplotlib for the survey data:
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
- Summary Statistics:
- Mean (Arithmetic Average): This provides a single value representing the "average" customer sentiment. For this dataset, the mean is $\approx 7.33$, indicating a generally positive lean.
- Median: The middle value when the data is sorted. It is less sensitive to outliers than the mean. Here, the median is $7.0$.
- Standard Deviation: This measures the amount of variation or dispersion. A lower standard deviation suggests customers have a consistent experience, while a higher one indicates polarized feedback. For this data, it is $\approx 1.58$.
- Mode: The most frequently occurring score. Identifying the mode helps pinpoint the most common customer response .
- Visualizations:
- Histogram: This is the most effective tool for seeing the "shape" of the data. It shows how many customers fall into each score bucket (1, 2, 3... 10), helping identify if the distribution is normal, skewed, or bimodal.
- Box Plot (Whisker Plot): Useful for identifying the range, the interquartile range , and any potential outliers .


```python
import matplotlib.pyplot as plt

# Dataset provided
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Create the histogram
# We set bins to range(1, 12) to ensure each integer 1-10 has its own centered bar
plt.hist(survey_scores, bins=range(1, 12), align='left', color='skyblue', edgecolor='black')

# Specific plot formatting
plt.title('Distribution of Customer Satisfaction Scores (1-10)')
plt.xlabel('Satisfaction Score')
plt.ylabel('Number of Customers')
plt.xticks(range(1, 11))  # Ensure all scale points are visible
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Save the visualization
plt.savefig('survey_distribution.png')
```

- Analysis of Results :
From the data, the distribution is slightly "left-skewed" (negatively skewed), meaning the bulk of the scores are on the higher end of the scale (7-9). This suggests that before the new product launch, the team has a solid foundation of satisfied customers, with very few reporting low satisfaction (below $6$).
